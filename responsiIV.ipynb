```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import cv2 as cv
import numpy as np


gambar = mpimg.imread("/content/sample_data/boy_color.gif")


if len(gambar.shape) == 3:
    gambar = cv.cvtColor(gambar, cv.COLOR_RGB2GRAY)

# Pastikan uint8
gambar = gambar.astype(np.uint8)

# Buat citra gelap dan cerah
gelap = cv.subtract(gambar, 50)
cerah = cv.add(gambar, 50)


T = 127
_, th_asli  = cv.threshold(gambar, T, 255, cv.THRESH_BINARY)
_, th_gelap = cv.threshold(gelap,  T, 255, cv.THRESH_BINARY)
_, th_cerah = cv.threshold(cerah,  T, 255, cv.THRESH_BINARY)

plt.figure(figsize=(12,8))

plt.subplot(3,3,1)
plt.title("Asli")
plt.imshow(gambar, cmap="gray")
plt.axis("off")

plt.subplot(3,3,2)
plt.title("Gelap")
plt.imshow(gelap, cmap="gray")
plt.axis("off")

plt.subplot(3,3,3)
plt.title("Cerah")
plt.imshow(cerah, cmap="gray")
plt.axis("off")

# ====== HASIL THRESHOLDING ======
plt.subplot(3,3,4)
plt.title("Threshold Asli")
plt.imshow(th_asli, cmap="gray")
plt.axis("off")

plt.subplot(3,3,5)
plt.title("Threshold Gelap")
plt.imshow(th_gelap, cmap="gray")
plt.axis("off")

plt.subplot(3,3,6)
plt.title("Threshold Cerah")
plt.imshow(th_cerah, cmap="gray")
plt.axis("off")

plt.tight_layout()
plt.show()
```

kode diatas ditujukan untuk menjalankan proses tresholding dengan hasil akhir membandingkan dengan citra sebelum dilakukan tresholding pada bagian atasnya

**perubahan kecerahan**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Baca gambar dalam grayscale
img = cv2.imread('/content/sample_data/boy_color.gif', cv2.IMREAD_GRAYSCALE)

if img is not None:
    # 2. Terapkan Gaussian Blur (mengurangi noise agar hasil sobel lebih bersih)
    img_blur = cv2.GaussianBlur(img, (3, 3), 0)

    # 3. Hitung Sobel Filter (Horizontal dan Vertikal)
    sobelx = cv2.Sobel(img_blur, cv2.CV_64F, 1, 0, ksize=3)
    sobely = cv2.Sobel(img_blur, cv2.CV_64F, 0, 1, ksize=3)

    # Gabungkan Magnitude (Hasil akhir deteksi tepi)
    sobel_filter = cv2.magnitude(sobelx, sobely)

    # PENTING: Konversi ke uint8 agar bisa dihitung histogramnya
    sobel_8bit = np.uint8(np.absolute(sobel_filter))

    # 4. Hitung Histogram
    hist_sobel_filter = cv2.calcHist([sobel_8bit], [0], None, [256], [0, 256])

    # 5. Menampilkan Visualisasi
    plt.figure(figsize=(12, 5))


    # Tampilkan Histogram Sobel
    plt.subplot(1, 2, 1)
    plt.plot(hist_sobel_filter, color='black')
    plt.title('Histogram of Sobel Filtered Image')
    plt.xlabel('Pixel Value')
    plt.ylabel('Frequency')
    plt.grid(alpha=0.3)

    hist_laplacian_filter = cv2.calcHist([lapcian_8bit], [0], None, [256], [0, 256])

    plt.subplot(1, 2, 2)
    plt.plot(hist_laplacian_filter)
    plt.title('Histogram of laplacian Filtered Image')
    plt.xlabel('Pixel Value')
    plt.ylabel('Frequency')

    plt.tight_layout()
    plt.show()
else:
    print("File gambar tidak ditemukan.")
```

Berdasarkan dari perbandinganya dapat dilihat bahwa nilai yang ada pada histogram laplacian lebih tinggi daripada yang ada pada histogram sobel
